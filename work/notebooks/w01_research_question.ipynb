## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

df = pd.read_csv('data/raw/content_refresh_anonymized.csv')

print("Rows, columns:", df.shape)
print(df['trend_direction'].value_counts())

df['is_down'] = df['trend_direction'] == 'down'

print("\n% of content trending down:", (df['is_down'].mean() * 100).round(1))

# avg_position: 0 means "no data" — exclude before computing
valid_pos = df[df['avg_position'] != 0]
print("\nAvg position, down vs not-down:")
print(valid_pos.groupby('is_down')['avg_position'].mean().round(1))

print("\nAvg CTR (%), down vs not-down:")
print(df.groupby('is_down')['ctr'].mean().round(2))
```
